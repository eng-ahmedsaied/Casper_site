```python
!pip install roboflow
from roboflow import Roboflow
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
import os
from google.colab import files

rf = Roboflow(api_key="UerUoO0p6dp4i37QCWxi")
project = rf.workspace("detection-fp8ei").project("leg-detection-ofgjl")
version = project.version(2)
dataset = version.download("yolov8")

model = YOLO("yolov8n.pt")
model.train(data=f"{dataset.location}/data.yaml", epochs=50, imgsz=640)

```

```python
output_folder = "runs/detect/predict"
os.makedirs(output_folder, exist_ok=True)

uploaded_files = files.upload()

for filename in uploaded_files.keys():
    results = model.predict(filename, save=True)
    print(f"Processed: {filename}")

    result_img_path = f"{output_folder}/{filename}"
    if os.path.exists(result_img_path):
        img = cv2.imread(result_img_path)
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        plt.imshow(img)
        plt.axis("off")
        plt.show()
```
